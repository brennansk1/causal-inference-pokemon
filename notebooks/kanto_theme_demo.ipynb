# Kanto Theme DemoA showcase of the custom matplotlib theme and plotting utilities that power this textbook.

```python
import sys, ossys.path.insert(0, os.path.abspath('../src'))import numpy as npimport pandas as pdimport matplotlib.pyplot as pltfrom kanto_utils import (    apply_kanto_theme, type_color, type_colors_dict, pokemon_scatter,    love_plot, did_plot, rdd_plot, badge_stamp,    oak_says, blue_says, nurse_joy_says, gym_leader_says,    badge_earned, blues_mistake,    load_trainers, load_cities_panel, load_happiness)apply_kanto_theme()np.random.seed(151)
```

## 1. Type Colors

```python
types = list(type_colors_dict().keys())colors = [type_color(t) for t in types]fig, ax = plt.subplots(figsize=(11, 3))for i, (t, c) in enumerate(zip(types, colors)):    ax.add_patch(plt.Rectangle((i, 0), 1, 1, color=c))    ax.text(i + 0.5, 0.5, t, ha='center', va='center', color='white', fontsize=9, fontweight='bold')ax.set_xlim(0, len(types))ax.set_ylim(0, 1)ax.set_xticks([])ax.set_yticks([])ax.set_title("Pokemon Type Palette")plt.tight_layout()plt.show()
```

## 2. Pokemon Scatter

```python
df = load_trainers()fig, ax = plt.subplots(figsize=(10, 6))pokemon_scatter(df.head(300), 'team_level_avg', 'badges_earned', hue='starter_type', ax=ax)ax.set_title("Trainer Team Level vs Badges Earned")plt.tight_layout()plt.show()
```

## 3. Love Plot (Balance Assessment)

```python
covs = ['badges', 'team_level_avg', 'experience', 'play_hours', 'wealth']smd_before = np.array([0.45, 0.38, 0.52, 0.31, 0.28])smd_after  = np.array([0.04, 0.02, 0.05, 0.03, 0.01])fig, ax = plt.subplots(figsize=(9, 5))love_plot(smd_before, smd_after, covs, ax=ax)ax.set_title("Love Plot: Covariate Balance Before/After Matching")plt.tight_layout()plt.show()
```

## 4. DiD Plot

```python
panel = load_cities_panel()fig, ax = plt.subplots(figsize=(10, 6))did_plot(panel, time_col='period', outcome_col='economic_index',         group_col='team_rocket_presence', treatment_period=8, ax=ax)ax.set_title("Team Rocket's Economic Impact (DiD Style)")plt.tight_layout()plt.show()
```
